## 11) Quadratic Equation Roots:

```python
import math
a = 1
b = 5
c = 6
discriminant = (b ** 2) - (4 * a * c)
root1=(-b + math.sqrt(discriminant)) / (2 * a)
root2=(-b - math.sqrt(discriminant)) / (2 * a)
print("The first root (Z1) is:",root1)
print("The second root (Z2) is:",root2)
```
