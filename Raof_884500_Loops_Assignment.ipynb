## Write a program to calculate the factorial of a number.
### Example:

* Input : 5
* Output 120

```python
n = int(input("Enter a non-negative integer: "))
if n < 0:
    print("Factorial is not defined for negative integers.")
else:
    factorial = 1
    for number in range(1, n + 1):
        factorial *= number
    print("Factorial:", factorial)
```

## Write a program to calculate:

* base ^ exponent
* without using **

```python
base = float(input("Enter base: "))
exponent = int(input("Enter integer exponent: "))

if base == 0 and exponent < 0:
    print("Zero cannot have a negative exponent.")
else:
    result = 1
    for _ in range(abs(exponent)):
        result *= base
    if exponent < 0:
        result = 1 / result
    print("Result:", result)
```

## 🏪 1. Shopping Cart System

### Goal:
    Create a shopping cart program that allows users to add items to their cart.
    Stop adding items when the user types "done".

### Steps:

    Ask the user to enter product names.
    
    Add each item to a list.
    
    If user types "done" → use break.
    
    Print all items added to the cart.

### Example:

    Enter item: Shirt  
    Enter item: Shoes  
    Enter item: done  
    Items in your cart: ['Shirt', 'Shoes']

```python
cart = []
total_bill = 0

while True:
    item_name = input("Enter item name (or type 'done' to finish): ")
    if item_name.lower() == "done":
        break
        
    price = float(input(f"Enter price for {item_name}: "))
    
    cart.append({"name": item_name, "price": price})
    total_bill += price

print("\n--- Shopping Cart receipt ---")
for item in cart:
    print(f"- {item['name']}: ${item['price']:.2f}")

print(f"\nTotal Bill: ${total_bill:.2f}")
```

## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]

```python
marks = []

while True:
    entry = int(input("Enter marks: "))
    
    if entry == -1:
        break
        
    if entry < 0 or entry > 100:
        print("Invalid marks")
        continue
        
    marks.append(entry)

if marks:
    average = sum(marks) / len(marks)
    print(f"Valid Marks: {marks}")
    print(f"Average = {average:.1f}")
else:
    print("No valid marks entered.")
```

### Goal:
    Simulate ordering from a fast food menu.

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0
account_created = False
account_name = ""

while True:
    print("\n--- BANKING SYSTEM ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")
    
    choice = input("Enter your choice (1-5): ")
    
    if choice == "1":
        account_name = input("Enter account holder name: ")
        balance = 0.0
        account_created = True
        print(f"Account created successfully for {account_name}!")
        
    elif choice == "2":
        if not account_created:
            print("Please create an account first!")
            continue
        amount = float(input("Enter amount to deposit: "))
        if amount > 0:
            balance += amount
            print(f"${amount:.2f} deposited successfully.")
        else:
            print("Invalid amount!")
            
    elif choice == "3":
        if not account_created:
            print("Please create an account first!")
            continue
        amount = float(input("Enter amount to withdraw: "))
        if amount > balance:
            print("Insufficient balance!")
        elif amount > 0:
            balance -= amount
            print(f"${amount:.2f} withdrawn successfully.")
        else:
            print("Invalid amount!")
            
    elif choice == "4":
        if not account_created:
            print("Please create an account first!")
            continue
        print(f"Account Holder: {account_name}")
        print(f"Current Balance: ${balance:.2f}")
        
    elif choice == "5":
        print("Thank you for using our banking system. Goodbye!")
        break
        
    else:
        print("Invalid choice! Please select between 1 and 5.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:

    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65

```python
all_totals = []

while True:
    student_name = input("\nEnter student name (or type 'done' to finish): ")
    if student_name.lower() == "done":
        break
        
    student_total = 0
    
    for j in range(3):
        marks = float(input(f"Subject {j + 1}: "))
        student_total += marks
        
    all_totals.append(student_total)

if all_totals:
    class_average = sum(all_totals) / len(all_totals)
    print(f"\n-> Output (Totals): {all_totals}")
    print(f"-> Class Average: {class_average:.2f}")
else:
    print("\nNo data entered.")
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
daily_totals = []

for day in range(1, 6):
    print(f"\nDay {day}")
    daily_total = 0

    for product in range(1, 4):
        sale = int(input(f"Enter sales for Product {product}: "))

        if sale == 0:
            continue

        daily_total += sale

    daily_totals.append(daily_total)

print("\n--- Sales Report ---")

for day in range(1, 6):
    print(f"Day {day}: {daily_totals[day - 1]}")

grand_total = sum(daily_totals)

print("Daily Totals:", daily_totals)
print("Grand Total:", grand_total)
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
for floor in range(1, 4):
    booked = 0
    available = 0
    print(f"Floor {floor}:")
    for room in range(1, 5):
        while True:
            answer = input(f"Room {room} booked? (Y/N): ").strip().upper()
            if answer in ("Y", "N"):
                break
            print("Please enter Y or N.")
        if answer == "Y":
            booked += 1
        else:
            available += 1
    print(f"Floor {floor}: {booked} booked, {available} available")
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]

```python
while True:
    invoice_name = input("\nEnter invoice name (or type 'done' to finish): ")
    if invoice_name.lower() == "done":
        break
        
    items = []
    subtotal = 0
    
    for j in range(4):
        price = float(input(f"Enter price for Item {j + 1}: "))
        items.append(price)
        subtotal += price
        
    tax = subtotal * 0.10
    total = subtotal + tax
    
    print(f"Items: {items}")
    print(f"-> Subtotal: {subtotal:.2f} | Tax (10%): {tax:.2f} | Total: {total:.2f}")
```

```python
while True:
    student_name = input("\nEnter student name (or type 'done' to finish): ")
    if student_name.lower() == "done":
        break
        
    total_marks = 0
    failed = False
    
    for j in range(4):
        marks = int(input(f"Enter marks for Subject {j + 1}: "))
        total_marks += marks
        
        if marks < 33:
            print(f"-> Failed in Subject {j + 1}")
            failed = True
            
    if not failed:
        average = total_marks / 4
        if average >= 80:
            grade = "A"
        elif average >= 60:
            grade = "B"
        elif average >= 33:
            grade = "C"
        else:
            grade = "F"
        print(f"-> Total Marks: {total_marks} | Grade: {grade}")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
while True:
    package_name = input("\nEnter package name (or type 'done' to finish): ")
    if package_name.lower() == "done":
        break
        
    total_rating = 0
    valid_features = 0
    
    for j in range(3):
        rating = float(input(f"Enter rating out of 5 for Feature {j + 1}: "))
        
        if rating == 0:
            continue
            
        total_rating += rating
        valid_features += 1
        
    if valid_features > 0:
        average = total_rating / valid_features
        print(f"-> Avg = {average:.2f}")
    else:
        print("-> No valid ratings entered for this package.")
```
