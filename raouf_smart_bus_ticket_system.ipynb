```python
##### SMART BUS TICKET BOOKING SYSTEM
# All data is stored in memory. Run this cell and use the menu below.

routes = [
    {
        "route_id": "R101",
        "departure_city": "Karachi",
        "destination_city": "Hyderabad",
        "departure_time": "08:00 AM",
        "ticket_price": 1200,
        "total_seats": 40,
        "available_seats": 40
    },
    {
        "route_id": "R102",
        "departure_city": "Karachi",
        "destination_city": "Sukkur",
        "departure_time": "09:30 AM",
        "ticket_price": 2000,
        "total_seats": 36,
        "available_seats": 36
    },
    {
        "route_id": "R103",
        "departure_city": "Hyderabad",
        "destination_city": "Mirpur Khas",
        "departure_time": "11:00 AM",
        "ticket_price": 800,
        "total_seats": 32,
        "available_seats": 32
    },
    {
        "route_id": "R104",
        "departure_city": "Karachi",
        "destination_city": "Lahore",
        "departure_time": "07:00 PM",
        "ticket_price": 5000,
        "total_seats": 45,
        "available_seats": 45
    },
    {
        "route_id": "R105",
        "departure_city": "Sukkur",
        "destination_city": "Hyderabad",
        "departure_time": "02:00 PM",
        "ticket_price": 1500,
        "total_seats": 30,
        "available_seats": 30
    }
]

# Tuple: supported departure cities
supported_cities = ("Karachi", "Hyderabad", "Sukkur", "Mirpur Khas", "Lahore")

# List of dictionaries: booking records
bookings = []

# Set: prevents duplicate booking IDs
issued_booking_ids = set()
next_booking_number = 1001

print("=" * 58)
print("       SMART BUS TICKET BOOKING & RESERVATION SYSTEM")
print("=" * 58)
print("Tip: Records stay in memory until the kernel is restarted.")

while True:
    print("\\n" + "-" * 58)
    print("MAIN MENU")
    print("1. View Available Routes")
    print("2. Search for a Route")
    print("3. Book a Ticket")
    print("4. View My Bookings")
    print("5. Cancel a Booking")
    print("6. View Seat Availability")
    print("7. Generate Booking Report")
    print("8. Exit")
    print("9. View Route Schedule")
    print("-" * 58)

    choice = input("Enter your choice (1-9): ").strip()

    if not choice.isdigit():
        print("Invalid input. Please enter a number from 1 to 9.")
        continue

    choice = int(choice)

    if choice == 1:
        show_available_only = input("Show only routes with seats available? (yes/no): ").strip().lower()
        print("\\nAVAILABLE BUS ROUTES")
        print("-" * 92)
        print(f"{'No.':<5}{'Route':<9}{'Journey':<35}{'Departure':<15}{'Fare (Rs.)':<13}{'Seats':<12}")
        print("-" * 92)

        for number, route in enumerate(routes, start=1):
            if show_available_only in ("yes", "y") and route["available_seats"] == 0:
                continue

            journey = route["departure_city"] + " -> " + route["destination_city"]
            if route["available_seats"] == 0:
                seat_status = "FULL"
            else:
                seat_status = str(route["available_seats"]) + "/" + str(route["total_seats"])

            print(f"{number:<5}{route['route_id']:<9}{journey:<35}{route['departure_time']:<15}{route['ticket_price']:<13}{seat_status:<12}")

    elif choice == 2:
        search_text = input("Search by route ID or city: ").strip().lower()
        if search_text == "":
            print("Search text cannot be empty.")
            continue

        print("\\nSEARCH RESULTS")
        found = False

        for route in routes:
            route_id = route["route_id"].lower()
            departure = route["departure_city"].lower()
            destination = route["destination_city"].lower()

            if search_text == route_id:
                print(route["route_id"], "|", route["departure_city"], "->", route["destination_city"],
                      "|", route["departure_time"], "| Rs.", route["ticket_price"],
                      "| Seats available:", route["available_seats"])
                found = True
                break
            elif search_text in departure or search_text in destination:
                print(route["route_id"], "|", route["departure_city"], "->", route["destination_city"],
                      "|", route["departure_time"], "| Rs.", route["ticket_price"],
                      "| Seats available:", route["available_seats"])
                found = True
        else:
            # for-else: runs when the loop finishes without break.
            # City searches can match several routes; the found flag handles that case.
            if not found:
                print("No route found for:", search_text)

    elif choice == 3:
        print("\\nBOOK A TICKET")
        passenger_name = input("Passenger name: ").strip()

        if passenger_name == "":
            print("Passenger name cannot be empty. Booking was not created.")
            continue

        age_text = input("Passenger age: ").strip()
        if not age_text.isdigit():
            print("Age must be a positive whole number. Booking was not created.")
            continue

        passenger_age = int(age_text)
        if passenger_age <= 0:
            print("Age must be greater than zero. Booking was not created.")
            continue

        route_id = input("Enter route ID (e.g. R101): ").strip().upper()
        selected_route = None

        for route in routes:
            if route["route_id"] == route_id:
                selected_route = route
                break

        if selected_route is None:
            print("Route ID not found. No booking was created.")
            continue

        seats_text = input("Number of seats: ").strip()
        if not seats_text.isdigit():
            print("Seat quantity must be a positive whole number. Booking was not created.")
            continue

        seats_requested = int(seats_text)
        if seats_requested <= 0:
            print("You must book at least one seat.")
            continue
        elif seats_requested > selected_route["available_seats"]:
            print("Not enough seats available. Available seats:", selected_route["available_seats"])
            continue
        else:
            # All checks passed: now create the booking and update seats.
            booking_id = "B" + str(next_booking_number)
            while booking_id in issued_booking_ids:
                next_booking_number += 1
                booking_id = "B" + str(next_booking_number)

            total_fare = selected_route["ticket_price"] * seats_requested

            booking = {
                "booking_id": booking_id,
                "passenger_name": passenger_name.title(),
                "passenger_age": passenger_age,
                "route_id": selected_route["route_id"],
                "seats": seats_requested,
                "total_fare": total_fare,
                "status": "Confirmed"
            }

            bookings.append(booking)
            issued_booking_ids.add(booking_id)
            selected_route["available_seats"] -= seats_requested
            next_booking_number += 1

            print("\\nBooking confirmed!")
            print("Booking ID:", booking_id)
            print("Passenger:", booking["passenger_name"])
            print("Age:", booking["passenger_age"])
            print("Route:", selected_route["route_id"], "(" + selected_route["departure_city"] + " -> " + selected_route["destination_city"] + ")")
            print("Seats booked:", seats_requested)
            print("Fare per seat: Rs.", selected_route["ticket_price"])
            print("Total fare: Rs.", total_fare)
            print("Status:", booking["status"])

    elif choice == 4:
        print("\\nBOOKING HISTORY")
        if len(bookings) == 0:
            print("No bookings found.")
        else:
            for number, booking in enumerate(bookings, start=1):
                print("\\nBooking", number)
                print("Booking ID:", booking["booking_id"])
                print("Passenger:", booking["passenger_name"], "| Age:", booking["passenger_age"])
                print("Route ID:", booking["route_id"])
                print("Seats:", booking["seats"])
                print("Total fare: Rs.", booking["total_fare"])
                print("Status:", booking["status"])

    elif choice == 5:
        booking_id_to_cancel = input("Enter booking ID to cancel: ").strip().upper()
        booking_found = False

        for booking in bookings:
            if booking["booking_id"] == booking_id_to_cancel:
                booking_found = True

                if booking["status"] == "Cancelled":
                    print("This booking has already been cancelled. Seats will not be restored again.")
                elif booking["status"] == "Confirmed":
                    for route in routes:
                        if route["route_id"] == booking["route_id"]:
                            route["available_seats"] += booking["seats"]
                            # Keep availability within the route capacity.
                            if route["available_seats"] > route["total_seats"]:
                                route["available_seats"] = route["total_seats"]
                            break

                    booking["status"] = "Cancelled"
                    print("Booking", booking_id_to_cancel, "has been cancelled.")
                    print("Seats restored:", booking["seats"])
                else:
                    print("This booking has an unknown status.")
                break
        else:
            print("Booking ID not found:", booking_id_to_cancel)

    elif choice == 6:
        available_only = input("Show only routes with seats remaining? (yes/no): ").strip().lower()
        print("\\nSEAT AVAILABILITY")
        for route in routes:
            if available_only in ("yes", "y") and route["available_seats"] == 0:
                continue

            if route["available_seats"] == route["total_seats"]:
                status = "Available"
            elif route["available_seats"] > 0:
                status = "Partially Booked"
            else:
                status = "Fully Booked"

            print(route["route_id"], "|", route["departure_city"], "->", route["destination_city"],
                  "|", route["available_seats"], "of", route["total_seats"], "seats |", status)

    elif choice == 7:
        total_bookings = len(bookings)
        confirmed_bookings = 0
        cancelled_bookings = 0
        reserved_seats = 0
        confirmed_revenue = 0
        fully_booked_routes = 0

        for booking in bookings:
            if booking["status"] == "Confirmed":
                confirmed_bookings += 1
                reserved_seats += booking["seats"]
                confirmed_revenue += booking["total_fare"]
            elif booking["status"] == "Cancelled":
                cancelled_bookings += 1

        for route in routes:
            if route["available_seats"] == 0:
                fully_booked_routes += 1

        print("\\nBOOKING REPORT")
        print("-" * 38)
        print("Total bookings created:", total_bookings)
        print("Confirmed bookings:", confirmed_bookings)
        print("Cancelled bookings:", cancelled_bookings)
        print("Seats currently reserved:", reserved_seats)
        print("Confirmed revenue: Rs.", confirmed_revenue)
        print("Fully booked routes:", fully_booked_routes)

        print("\\nCONFIRMED BOOKINGS BY ROUTE")
        for route in routes:
            route_booking_count = 0
            route_seats = 0
            route_revenue = 0

            for booking in bookings:
                if booking["route_id"] == route["route_id"] and booking["status"] == "Confirmed":
                    route_booking_count += 1
                    route_seats += booking["seats"]
                    route_revenue += booking["total_fare"]

            print(route["route_id"], "|", route["departure_city"], "->", route["destination_city"],
                  "| Bookings:", route_booking_count, "| Seats:", route_seats,
                  "| Revenue: Rs.", route_revenue)

    elif choice == 8:
        print("Thank you for using Smart Bus Ticket Booking System. Goodbye!")
        break

    elif choice == 9:
        print("\\nROUTE SCHEDULE BY DEPARTURE CITY")
        for city in supported_cities:
            print("\\n" + city)
            departure_found = False

            for route in routes:
                if route["departure_city"].lower() == city.lower():
                    print(" ", route["route_id"], "->", route["destination_city"],
                          "|", route["departure_time"])
                    departure_found = True

            if not departure_found:
                print("  No scheduled departures.")

    else:
        print("Unsupported choice. Please select a number from 1 to 9.")
        continue
```
