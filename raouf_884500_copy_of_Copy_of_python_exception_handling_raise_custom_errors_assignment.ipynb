### Q13. Student Record Registration

### Q14. Weak Password Rejection

A website requires passwords to contain at least 8 characters, including at least one digit. A password that does not meet the rules must be rejected.

Create a custom exception named `WeakPasswordError`. Ask the user for a password, check both rules, raise the custom exception when a rule fails, and handle it to explain what needs to be corrected. Do not display the actual password in the success message.

```python
class WeakPasswordError(Exception):
    pass

try:
    password = input("Enter password: ")

    if len(password) < 8:
        raise WeakPasswordError("Password must contain 8 characters")

    if not any(char.isdigit() for char in password):
        raise WeakPasswordError("Password must contain a digit")

    print("Password accepted")

except WeakPasswordError as e:
    print("Error:", e)
```

### Q22. Bank Transfer Between Two Accounts

A banking system transfers money from one account to another. The transfer amount must be positive, the sender and receiver must be different accounts, and the sender must have enough funds.

Create custom exceptions named `InvalidTransferError` and `InsufficientFundsError`. Validate all rules and raise the correct exception. Handle the exceptions and ensure that neither account balance changes when any validation fails. Display both balances after the transaction attempt.

```python
class InvalidTransferError(Exception):
    pass

class InsufficientFundsError(Exception):
    pass

account1 = 10000
account2 = 5000

try:
    amount = float(input("Enter transfer amount: "))

    if amount <= 0:
        raise InvalidTransferError("Amount must be positive")

    if account1 == account2:
        raise InvalidTransferError("Accounts must be different")

    if amount > account1:
        raise InsufficientFundsError("Not enough funds")

    account1 -= amount
    account2 += amount

    print("Transfer successful")

except InvalidTransferError as e:
    print(e)

except InsufficientFundsError as e:
    print(e)

except ValueError:
    print("Invalid amount")

finally:
    print("Account 1:", account1)
    print("Account 2:", account2)
```

### Q26. Ride-Hailing Fare and Payment

### Q29. Inventory Purchase with Safe Transaction

A warehouse purchase operation receives a product ID and requested quantity. The product must exist, the quantity must be positive, and enough stock must be available. A failed purchase must leave inventory unchanged.

Create custom exceptions named `ProductNotFoundError`, `InvalidQuantityError`, and `InsufficientStockError`. Implement validation before changing stock. Use `try`, `except`, `else`, and `finally` meaningfully: report success only in `else`, handle expected failures in `except`, and display an audit/attempt-finished message in `finally`. Demonstrate that stock remains unchanged after a failed purchase.

```python
class ProductNotFoundError(Exception):
    pass

class InvalidQuantityError(Exception):
    pass

class InsufficientStockError(Exception):
    pass

inventory = {
    "P101": 20,
    "P102": 10,
    "P103": 5
}

try:
    product_id = input("Enter product ID: ")
    quantity = int(input("Enter quantity: "))

    if product_id not in inventory:
        raise ProductNotFoundError("Product not found")

    if quantity <= 0:
        raise InvalidQuantityError("Quantity must be positive")

    if quantity > inventory[product_id]:
        raise InsufficientStockError("Not enough stock")

except ProductNotFoundError as e:
    print(e)

except InvalidQuantityError as e:
    print(e)

except InsufficientStockError as e:
    print(e)

except ValueError:
    print("Invalid quantity")

else:
    inventory[product_id] -= quantity
    print("Purchase successful")

finally:
    print("Purchase attempt finished")
    print("Current inventory:", inventory)
```

### Q30. Mini Banking System with a Custom Exception Hierarchy

Design a menu-based mini banking program that supports deposit, withdrawal, and balance viewing. Deposits and withdrawals must be positive. A withdrawal cannot exceed the available balance. Invalid menu choices must also be handled.

Create a base custom exception named `BankingError`, then create `InvalidAmountError` and `InsufficientBalanceError` as child classes of `BankingError`. Raise the appropriate exception from the relevant operation and handle the base class or specific child classes safely. Keep the program running after an invalid transaction, and show the final account balance when the user chooses to exit.

```python
class BankingError(Exception):
    pass

class InvalidAmountError(BankingError):
    pass

class InsufficientBalanceError(BankingError):
    pass

balance = 10000

while True:

    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    try:
        choice = int(input("Enter choice: "))

        if choice == 1:

            amount = float(input("Enter deposit: "))

            if amount <= 0:
                raise InvalidAmountError("Invalid deposit amount")

            balance += amount
            print("Deposit successful")

        elif choice == 2:

            amount = float(input("Enter withdrawal: "))

            if amount <= 0:
                raise InvalidAmountError("Invalid withdrawal amount")

            if amount > balance:
                raise InsufficientBalanceError("Insufficient balance")

            balance -= amount
            print("Withdrawal successful")

        elif choice == 3:

            print("Current balance:", balance)

        elif choice == 4:

            print("Final balance:", balance)
            print("Thank you for using our bank")
            break

        else:
            print("Invalid menu choice")

    except BankingError as e:
        print("Banking error:", e)

    except ValueError:
        print("Please enter valid numbers")
```
